# Plots

```python
fig, ax = plt.subplots(figsize=(3.375, 2.0))

phi_target = -(E_FCI - h3) * t
phi_target_signed = (phi_target + np.pi) % (2 * np.pi) - np.pi

ax.axvline(
    phi_target_signed,
    linestyle="--",
    linewidth=0.9,
    color="red"
)

ax.plot(
    x_axis1,
    y_axis1,
    linewidth=1.0
)

ax.set_xlabel(r"$\tilde{\phi}$", fontsize=9)
ax.set_ylabel(r"$\log Q(\tilde{\phi})$", fontsize=9)

ax.set_ylim(-4_000, 0)

ax.tick_params(which="both", direction="in", top=False, right=False)
ax.tick_params(which="major", length=3.0, width=0.8, labelsize=8)
ax.tick_params(which="minor", length=1.8, width=0.6)

fig.subplots_adjust(
    left=0.18,
    right=0.98,
    bottom=0.18,
    top=0.96
)

plt.show()
```
